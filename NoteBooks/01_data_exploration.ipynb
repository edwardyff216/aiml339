```python
chat_with_slot = chat.copy()

def convert_player_slot(slot):
    if 0 <= slot <= 4:
        return slot
    elif 5 <= slot <= 9:
        return slot + 123
    else:
        return pd.NA

chat_with_slot["player_slot"] = (
    chat_with_slot["slot"]
    .apply(convert_player_slot)
    .astype("Int64")
)

display(
    chat_with_slot[
        ["match_id", "slot", "player_slot", "unit", "key"]
    ].head(20)
)
```

```python
# Pick one real chat message from match 0
sample_message = (
    chat_with_slot[
        (chat_with_slot["match_id"] == 0) &
        (chat_with_slot["time"] >= 0)
    ]
    .sort_values("time")
    .iloc[-1]
)

print("Message:")
print(sample_message["key"])

print("\nMatch ID:", sample_message["match_id"])
print("Chat time:", sample_message["time"])
print("Player slot:", sample_message["player_slot"])
```

```python
match_id = sample_message["match_id"]
chat_time = sample_message["time"]
player_slot = int(sample_message["player_slot"])

# Get all time snapshots from this match
match_time_data = player_time[
    player_time["match_id"] == match_id
]

# Only use information available at or before the chat message
previous_states = match_time_data[
    match_time_data["times"] <= chat_time
]

state = previous_states.sort_values("times").iloc[-1]

print("Chat time:", chat_time)
print("Context snapshot time:", state["times"])

print("\nGame context:")
print("Gold:", state[f"gold_t_{player_slot}"])
print("XP:", state[f"xp_t_{player_slot}"])
print("Last Hits:", state[f"lh_t_{player_slot}"])
```

```python
def get_game_context(message_row, player_time_df):
    match_id = message_row["match_id"]
    chat_time = message_row["time"]
    player_slot = message_row["player_slot"]

    # 无法识别玩家的聊天
    if pd.isna(player_slot):
        return None

    # 暂时不处理开局前聊天
    if chat_time < 0:
        return None

    player_slot = int(player_slot)

    # 找这一场比赛
    match_data = player_time_df[
        player_time_df["match_id"] == match_id
    ]

    # 只允许使用发言时或发言前的数据
    previous_states = match_data[
        match_data["times"] <= chat_time
    ]

    if previous_states.empty:
        return None

    # 最近一个历史 snapshot
    state = previous_states.sort_values("times").iloc[-1]

    return {
        "snapshot_time": state["times"],
        "gold": state[f"gold_t_{player_slot}"],
        "xp": state[f"xp_t_{player_slot}"],
        "last_hits": state[f"lh_t_{player_slot}"]
    }
```

```python
context = get_game_context(
    sample_message,
    player_time
)

print(context)
```

```python
test_messages = (
    chat_with_slot[
        (chat_with_slot["time"] >= 0) &
        (chat_with_slot["player_slot"].notna())
    ]
    .head(10)
)

for _, row in test_messages.iterrows():
    context = get_game_context(row, player_time)

    print("Message:", row["key"])
    print("Time:", row["time"])
    print("Player slot:", row["player_slot"])
    print("Context:", context)
    print("-" * 50)
```

```python
chat_context = chat_with_slot.copy()

# 暂时只保留正式比赛开始后的、能对应玩家的聊天
chat_context = chat_context[
    (chat_context["time"] >= 0) &
    (chat_context["player_slot"].notna())
].copy()

# 找“发言之前最近一个60秒快照”
chat_context["snapshot_time"] = (
    chat_context["time"] // 60
) * 60

print("Rows:", len(chat_context))

display(
    chat_context[
        ["match_id", "time", "snapshot_time",
         "player_slot", "unit", "key"]
    ].head(20)
)
```

```python
PLAYER_SLOTS = [0, 1, 2, 3, 4, 128, 129, 130, 131, 132]

player_time_long_parts = []

for slot in PLAYER_SLOTS:
    temp = player_time[
        [
            "match_id",
            "times",
            f"gold_t_{slot}",
            f"xp_t_{slot}",
            f"lh_t_{slot}"
        ]
    ].copy()

    temp["player_slot"] = slot

    temp = temp.rename(
        columns={
            f"gold_t_{slot}": "gold",
            f"xp_t_{slot}": "xp",
            f"lh_t_{slot}": "last_hits"
        }
    )

    player_time_long_parts.append(temp)

player_time_long = pd.concat(
    player_time_long_parts,
    ignore_index=True
)

print("Shape:", player_time_long.shape)

display(player_time_long.head(20))
```

```python
key_columns = ["match_id", "times", "player_slot"]

duplicates = player_time_long.duplicated(
    subset=key_columns
).sum()

print("Duplicate context keys:", duplicates)
```

```python
player_time_lookup = player_time_long.rename(
    columns={"times": "snapshot_time"}
)

chat_game_context = chat_context.merge(
    player_time_lookup,
    on=[
        "match_id",
        "snapshot_time",
        "player_slot"
    ],
    how="left",
    validate="many_to_one"
)

print("Original chat rows:", len(chat_context))
print("Merged rows:", len(chat_game_context))

display(
    chat_game_context[
        [
            "match_id",
            "time",
            "snapshot_time",
            "player_slot",
            "unit",
            "key",
            "gold",
            "xp",
            "last_hits"
        ]
    ].head(20)
)
```

```python
print("Missing context values:")

print(
    chat_game_context[
        ["gold", "xp", "last_hits"]
    ].isna().sum()
)

matched = chat_game_context["gold"].notna().sum()
total = len(chat_game_context)

print("\nMatched:", matched)
print("Total:", total)
print("Match rate:", matched / total)
```

```python
time_difference = (
    chat_game_context["time"]
    - chat_game_context["snapshot_time"]
)

print("Minimum time difference:", time_difference.min())
print("Maximum time difference:", time_difference.max())

print(
    "All snapshots are at or before message:",
    (time_difference >= 0).all()
)
```

```python
# Preserve the original order of messages
chat_game_context["message_order"] = np.arange(len(chat_game_context))

# Sort messages inside each match
chat_game_context = chat_game_context.sort_values(
    ["match_id", "time", "message_order"]
).reset_index(drop=True)

# Previous 5 chat messages in the same match
for i in range(1, 6):
    chat_game_context[f"prev_message_{i}"] = (
        chat_game_context
        .groupby("match_id")["key"]
        .shift(i)
    )

display(
    chat_game_context[
        [
            "match_id",
            "time",
            "unit",
            "key",
            "prev_message_1",
            "prev_message_2",
            "prev_message_3",
            "prev_message_4",
            "prev_message_5"
        ]
    ].head(20)
)
```

```python
first_messages = (
    chat_game_context
    .groupby("match_id")
    .head(1)
)

print("Number of matches:", len(first_messages))

print(
    "First message has previous context:",
    first_messages["prev_message_1"].notna().sum()
)
```

```python
PROCESSED_DIR = PROJECT_ROOT / "data" / "processed"
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

base_output_path = PROCESSED_DIR / "chat_game_base.parquet"

chat_game_context.to_parquet(
    base_output_path,
    index=False
)

print("Saved to:")
print(base_output_path)
```
